# Tutorial about configuration managment

Locan makes use of configuration settings as defined in the class locan.configuration.Configuration.

Pydantic-settings is used for type checking and validation.

A configuration is instantiated upon locan startup from either (in order of precedence):
    * [tool.locan] settings in pyproject.toml,
    * a locan_config.toml file,
    * a .env file,
    * environment variables,
    * or default settings.

```python
import tempfile
from pathlib import Path

from pydantic import ValidationError

import locan as lc
```

```python
lc.show_versions(system=False, dependencies=False, verbose=False)
```

## Default configuration

When importing locan, a configuration class is instantiated from default values, environment variables, or from a configuration file by searching through the following file hirarchy (depending on the current platform):

```python
lc.configuration.CONFIGURATION_FILE_HIERARCHY
```

You can check if a configuration file is found and get its path. If no file was found None is returned.

```python
lc.configuration.get_configuration_file_path()
```

The configuration can be accessed and viewed through:

```python
lc.config
```

```python
lc.config.asdict()
```

```python
lc.config.data_directory
```

The configuration can be altered programmatically:

```python
lc.config.data_directory = Path(".").absolute()
lc.config.data_directory
```

## Configuration settings

Details about the configuration settings are found in the Configuration class docstring:

```python
print(lc.config.__doc__)
```

A config.toml file with all default values can be saved and customized.

```python
file_path = tempfile.NamedTemporaryFile(suffix=".toml").name
lc.config.to_toml(path=file_path)

lc.config = lc.Configuration.from_toml(file_path=file_path)
lc.config
```

## Custom configuration

A custom configuration can always be instantiated from the `Configuration` class:

```python
config = lc.Configuration(
    name='MyLocanConfiguration',
    data_directory=Path(".").absolute(),
)
config
```

or from a custom configuration file:

```python
config_file_path = Path(".") / "my_locan_config.toml"
config = lc.Configuration.from_toml(file_path=config_file_path)
config
```

You can save a toml-file for any Configuration instance:

```python
file_path = tempfile.NamedTemporaryFile(suffix=".toml").name
config.to_toml(path=file_path)
```

Custom fields are not allowed unless a new Configuration class is defined.

```python
try:
    config = lc.Configuration(
        custom_field="my custom value"
    )
except ValidationError as exception:
    print(exception)
```

```python
class MyConfiguration(lc.Configuration):
    custom_field: str = "my default value"

config = MyConfiguration(
    custom_field="my custom value"
)
config
```
